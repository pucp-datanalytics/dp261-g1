```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.config import *
from src.io_utils import load_kick_data, save_df
from src.preprocessing import prepare_features, split_X_y

pd.set_option('display.max_columns', 120)
print('Proyecto:', PROJECT_ROOT)

from ipywidgets import interact, Dropdown, IntSlider
```

```python
df = load_kick_data()
numeric_cols = df.select_dtypes(include='number').columns.drop(TARGET, errors='ignore').tolist()
cat_cols = df.select_dtypes(include='object').columns.tolist()
```

```python
@interact(variable=Dropdown(options=numeric_cols, description='Variable'), bins=IntSlider(value=30, min=10, max=80, step=5))
def plot_numeric(variable, bins):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    sns.histplot(data=df, x=variable, hue=TARGET, bins=bins, ax=axes[0], stat='density', common_norm=False)
    sns.boxplot(data=df, x=TARGET, y=variable, ax=axes[1])
    axes[0].set_title(f'Distribución de {variable}')
    axes[1].set_title(f'{variable} por target')
    plt.tight_layout()
    plt.show()
```

```python
@interact(variable=Dropdown(options=cat_cols, description='Variable'))
def plot_categorical(variable):
    tab = df.groupby(variable, dropna=False)[TARGET].agg(['count', 'mean']).sort_values('count', ascending=False).head(15)
    display(tab)
    fig, ax = plt.subplots(figsize=(10, 4))
    tab['mean'].sort_values().plot(kind='barh', ax=ax)
    ax.set_title(f'Tasa de mala compra por {variable}')
    ax.set_xlabel('Tasa IsBadBuy=1')
    plt.show()
```
