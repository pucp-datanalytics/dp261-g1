## 1. Por qué la API debe cargar el pipeline completo

El usuario no va a subir variables como `odo_per_year` o `warranty_to_cost`. El usuario sube las columnas crudas del dataset original.

Por eso `models/final_model.pkl` contiene:

```text
FeatureEngineeringTransformer -> preprocessor -> model
```

La API no debe replicar manualmente lógica de features. Solo carga el pipeline y predice.
